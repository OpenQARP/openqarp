### Preparing a circuit to plot

```python
from qarp.blocks import HypergraphStateBlock, CompositeBlock
import numpy as np
```

```python
block1 = HypergraphStateBlock(edges=[(0, 1), (2, 3), (1, 2, 3)], n_qubits=4, name="Block1").build()
block1.sdg(2)
block1.rz(1, 0.12345 * np.pi)
block1.cswap(3, 2, 1)
block1.x(2)
block1.measure([(q, q) for q in range(block1.n_qubits)])
block1.plot(interactive=False)

block2 = CompositeBlock([block1, block1]).build()
block2.plot()
```

### Plot arguments and features

The arguments in input are the following.

- `circ`: The quantum circuit to plot.
- `color_scheme`: Color scheme for gates ('default', 'colorblind', or custom dict).
- `figsize`: Size of the figure (width, height). If None, size is auto-calculated.
- `save_fig`: If provided, saves the figure to the given file path.
- `decompose_boxes`: Whether to decompose box gates before plotting.
- `flatten_layers`: Whether to flatten layers in the circuit.
- `invert_order`: Whether to invert the order of qubits in the plot.
- `scrollable`: If True, display the plot in a scrollable HTML container (for Jupyter).
- `use_latex`: Whether to use LaTeX for rendering text.
- `spacing`: Horizontal spacing between gates. If None, uses default from config.
- `verbose`: If True, prints additional information about the circuit.
- `label_mode`: How to handle gate labels. Options:
    - `LabelMode.TRUNCATE`: Show only gate names without parameters (safest, avoids overlaps)
    - `LabelMode.SMART`: Intelligent positioning to avoid overlaps (default)
    - `LabelMode.FULL`: Full labels without adjustment (may overlap)
- `show_gate_indices`: If True, display small index numbers near each gate.
- `return_plotter`: If True, returns the plotter instance for gate querying via `query_gate(index)` or `list_gates()`.

**Plotter methods for inspecting gates:**
- `plotter.query_gate(index)`: Get full information about a gate by its index.
- `plotter.list_gates()`: Print a summary of all gates.
- `plotter.get_inner_circuit(index)`: Extract the inner circuit from a boxed sub-block (plain or controlled).
- `plotter.plot_inner(index, **kwargs)`: Plot the inner circuit of a box gate directly.

```python
block1.plot( color_scheme='default', decompose_boxes=False, figsize=None, save_fig=None, flatten_layers=True, invert_order=False, scrollable=False, use_latex=False, spacing=0.25, verbose=False)
```

```python
plotter = block2.plot(show_gate_indices=True, return_plotter=True)

# List all gates
plotter.list_gates()

print()

# Query a specific gate for full information
gate = plotter.query_gate(1)
print(f"Gate type: {gate.gate_type}")
print(f"Full label: {gate.full_label}")
print(f"Qubits: {gate.qubits}")

plotter.plot_inner(1)
```
