```python
from qarp.operators import QubitOperator
from qarp.factories import PauliBlockFactory
```

```python
op = QubitOperator("X0 Y1 Z2")

factory = PauliBlockFactory().create_blocks(op, change_basis=True, measure=True)

print(factory[0].coefficient, factory[0].pauli_string)

factory[0].build().plot()
```

```python
# Define a QubitOperator with multiple terms
op = 1/2 * QubitOperator('X0 Y1 Z3') + 2 * QubitOperator('Z0 Z1') + 0.5 * QubitOperator('Y1 X2') + 3 * QubitOperator('Z2')

# Create Pauli blocks from the operator using the factory
factory = PauliBlockFactory().create_blocks(op)

# Iterate through the generated Pauli blocks showing their coefficients, Pauli strings, and plots
for i, pauli_block in enumerate(factory):
    print(f"Pauli Block [{i}] :", pauli_block.coefficient, pauli_block.pauli_string)
    pauli_block.build().plot()
```
