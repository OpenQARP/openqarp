# Grover search with a known marked count

This minimal example searches four basis states and marks integer label 2. With one marked state, one Grover iteration succeeds with probability one. The construction follows [Grover's original search algorithm](https://arxiv.org/abs/quant-ph/9605043) and the known-solution-count analysis of [Boyer, Brassard, Høyer, and Tapp](https://arxiv.org/abs/quant-ph/9605034).

```python
import numpy as np

import qarp
from qarp.algorithms import Grover
from qarp.algorithms import Sampler
from qarp.blocks import SimpleBlock
```

## Mark label 2

OpenQARP uses LSB-first qubits, so label 2 has bits `(0, 1)`. X-conjugating qubit 0 maps that basis state to $|11\rangle$ for the multi-controlled Z phase flip. The resulting oracle implements exactly $I-2|2\rangle\langle2|$.

```python
oracle = SimpleBlock(2, name="O_good")
oracle.x(0)
oracle.mcz([0, 1])
oracle.x(0)
```

## Run Grover search

`n_marked=1` selects the iteration count. `good_states=[2]` is optional reporting metadata; it does not construct or change the oracle.

```python
algorithm = Grover(
    oracle,
    n_marked=1,
    good_states=[2],
    primitive=Sampler(n_shots=qarp.EXACT),
).build()
distribution = algorithm.run()

assert algorithm.n_iterations == 1
assert algorithm.most_likely_states == (2,)
assert np.isclose(algorithm.predicted_success_probability, 1.0, atol=1e-12)
assert np.isclose(algorithm.success_probability, 1.0, atol=1e-12)

print("Distribution:", distribution)
print("Iterations:", algorithm.n_iterations)
print("Most likely states:", algorithm.most_likely_states)
print("Sampled success probability:", algorithm.success_probability)
```
